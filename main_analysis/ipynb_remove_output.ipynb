```python
from nbformat import read, write

def remove_output(book):
    for cell in book.cells:
        if hasattr(cell, "outputs"):
            cell.outputs = []
        if hasattr(cell, "prompt_number"):
            del cell["prompt_number"]

book = read(open(r"C:\Users\raznu\Development\RESEARCH\MY_CODES\Raanana_urban_floods\SWMM_Calibration/SWMM_Cross_Validation.ipynb"), 4)
remove_output(book)
write(book, open(r"C:\Users\raznu\Development\RESEARCH\MY_CODES\Raanana_urban_floods\SWMM_Calibration/SWMM_Cross_Validation.ipynb", "w"), 4)
```
