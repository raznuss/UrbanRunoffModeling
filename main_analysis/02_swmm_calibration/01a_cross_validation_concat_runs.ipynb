```python
import os
import pickle
import pandas as pd

path = r'D:\Development\RESEARCH\Raanana\SWMM/'

# List to store all dictionaries
all_dicts = []

# Iterate through potential filenames until a FileNotFoundError is raised
num_run = 1
while True:
#     filename = os.path.join(path, f'cross_validation_cali_dict_28catchments_part_{num_run}.pickle')
    filename = os.path.join(path, f'cross_validatio_cali_dict_28catchments_part_{num_run}.pickle')
    try:
        with open(filename, 'rb') as file:
            cross_validation_dict = pickle.load(file)

            # Append the loaded dictionary to the list
            all_dicts.append(cross_validation_dict)

    except FileNotFoundError:
        break  # No more files found

    num_run += 1  # Increment for the next potential filename

# New dictionary to store the concatenated dataframes
cross_validation_complete_dict = {}

# Iterate through keys in the dictionaries
for key in all_dicts[0].keys():
    # Concatenate the dataframes from all dictionaries for each key
    concatenated_df0 = pd.concat([d[key][0] for d in all_dicts])
    concatenated_df1 = pd.concat([d[key][1] for d in all_dicts])

    # Reset index for concatenated dataframes
    concatenated_df0.reset_index(drop=True, inplace=True)
    concatenated_df1.reset_index(drop=True, inplace=True)

    # Save only one of the third dataframes
    df3 = all_dicts[0][key][2]  # Assuming all dictionaries have the same structure

    # Store the concatenated and third dataframes in the new dictionary
    cross_validation_complete_dict[key] = [concatenated_df0, concatenated_df1, df3]

# Save the result dictionary as a pickle file in the same path
output_filename = os.path.join(path, 'cross_validation_complete_dict_28.pickle')
with open(output_filename, 'wb') as file:
    pickle.dump(cross_validation_complete_dict, file)

# cross_validation_complete_dict is now saved as a pickle file in the specified path
```
