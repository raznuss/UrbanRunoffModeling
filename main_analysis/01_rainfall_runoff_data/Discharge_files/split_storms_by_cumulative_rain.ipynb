```python
def find_sim_time_range(rain_serie, time_window_size=6, rainfall_threshold=5, minimum_range_time=12):
    interval = int(time_window_size * 60 / 5)  # convert to 5-minute intervals
    minimum_time_range = int(minimum_range_time * 60 / 5)
    sim_start_time = None
    sim_end_time = None
    start_timestamp = None
    end_timestamp = None

    rain_df = pd.DataFrame(rain_serie, columns=['rainfall [mm/h]'])  # Create a DataFrame
    rain_df.reset_index(inplace=True)

    for index, row in rain_df.iterrows():
        start_index = max(0, index - interval + 1)
        end_index = index + 1

        if end_index >= len(rain_df):  # Check if end_index is out of bounds
            break  # Exit the loop to avoid the IndexError

        cumulative_rain = rain_df.loc[start_index:end_index, 'rainfall [mm/h]'].sum()

        if cumulative_rain > rainfall_threshold and sim_start_time is None:
            sim_start_time = rain_df.loc[start_index]['Time']
            start_timestamp = pd.to_datetime(rain_df.iloc[start_index]['Time'])

        if cumulative_rain < rainfall_threshold and sim_start_time is not None:
            end_timestamp = pd.to_datetime(rain_df.iloc[end_index]['Time'])
            time_range_hours = (end_timestamp - start_timestamp).total_seconds() / 3600  # Convert to hours

            if time_range_hours >= minimum_range_time:
                sim_end_time = rain_df.loc[end_index]['Time']
                break  # Exit the loop when the condition is met

    return sim_start_time, sim_end_time
```

```python
for index, cross_val in enumerate(cross_validation_dict.keys()):
    rain_serie = cross_validation_dict[cross_val][7]['rainfall [mm/h]']
    start_time, end_time = find_sim_time_range(rain_serie, time_window_size=6, rainfall_threshold=6, minimum_range_time=0)
    
    print(cross_val)
    print(f"Simulation Start Time: {start_time}")
    print(f"Simulation End Time: {end_time}\n\n")  # Use '\n\n' for double newlines
```
