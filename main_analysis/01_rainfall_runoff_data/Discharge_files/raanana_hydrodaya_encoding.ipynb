### encoding data 

```python
import os
import pandas as pd
import datetime
from datetime import datetime
```

```python
## encoding hydro data csv file whith jeibrish 
directory = r'C:\Users\raznu\Development\RESEARCH\Raanana\data\hydrodata/'
columns = ['date_and_time', 'discharge_cms']
encoding = 'latin-1'

for filename in os.listdir(directory):
#     print(filename)
    if filename.endswith(".csv"):
#         df = pd.read_csv(directory + filename, header = None , names = columns)
        df = pd.read_csv(directory + filename, header=None, names=columns, encoding=encoding)
        df = df.drop(df.index[[0,1]])
        df = df.reset_index(drop=True)
        date_and_time = pd.to_datetime(df['date_and_time'][0] , format='%Y_%m_%d', errors='ignore')
        date = str(pd.to_datetime(date_and_time).date())
        date = str(date_and_time[0:10])
        date = date.replace("/","_") ; date = date.split("_") ; date = "_".join((date[2], date[1] , date[0])) 
        df.to_csv(directory + date + '.csv')
    else:
        continue
```
